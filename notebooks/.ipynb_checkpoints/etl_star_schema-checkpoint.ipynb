```python

from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.window import Window
from pyspark.sql.types import IntegerType, FloatType, StringType, DateType
import psycopg2

JDBC_URL = "jdbc:postgresql://postgres:5432/postgres"
DB_PROPS = {
    "user": "postgres",
    "password": "POSTGRES_PASSWORD",
    "driver": "org.postgresql.Driver"
}
JAR_PATH = "/home/jovyan/jars/postgresql-42.7.4.jar"

spark = SparkSession.builder \
    .appName("ETL Star Schema") \
    .config("spark.jars", JAR_PATH) \
    .getOrCreate()
print("Spark version:", spark.version)
```

```python
csv_files = "/home/jovyan/data/MOCK_DATA*.csv"
df_raw = spark.read.csv(csv_files, header=True, inferSchema=True)

df_raw = df_raw.withColumn("sale_date", 
                           when(col("sale_date") == "NULL", None).otherwise(col("sale_date")))
df_raw = df_raw.withColumn("sale_date", to_date(col("sale_date"), "M/d/yyyy"))
df_raw = df_raw.withColumn("product_release_date", to_date(col("product_release_date"), "M/d/yyyy"))
df_raw = df_raw.withColumn("product_expiry_date", to_date(col("product_expiry_date"), "M/d/yyyy"))

df_raw = df_raw.withColumn("customer_age", col("customer_age").cast(IntegerType()))
df_raw = df_raw.withColumn("product_price", col("product_price").cast(FloatType()))
df_raw = df_raw.withColumn("product_weight", col("product_weight").cast(FloatType()))
df_raw = df_raw.withColumn("product_rating", col("product_rating").cast(FloatType()))
df_raw = df_raw.withColumn("product_reviews", col("product_reviews").cast(IntegerType()))
df_raw = df_raw.withColumn("sale_quantity", col("sale_quantity").cast(IntegerType()))
df_raw = df_raw.withColumn("sale_total_price", col("sale_total_price").cast(FloatType()))

df_raw.cache()
print(f"Всего строк: {df_raw.count()}")
df_raw.select("sale_date").show(5)
```

```python

df_dates = df_raw.select("sale_date").distinct().na.drop()
df_dim_date = df_dates.withColumn("date_id", date_format(col("sale_date"), "yyyyMMdd").cast("int")) \
    .withColumn("full_date", col("sale_date")) \
    .withColumn("year", year(col("sale_date")).cast("smallint")) \
    .withColumn("quarter", quarter(col("sale_date")).cast("smallint")) \
    .withColumn("month", month(col("sale_date")).cast("smallint")) \
    .withColumn("month_name", date_format(col("sale_date"), "MMMM")) \
    .withColumn("day", dayofmonth(col("sale_date")).cast("smallint")) \
    .withColumn("day_of_week", dayofweek(col("sale_date")).cast("smallint")) \
    .withColumn("day_name", date_format(col("sale_date"), "EEEE")) \
    .withColumn("week_of_year", weekofyear(col("sale_date")).cast("smallint")) \
    .drop("sale_date")
df_dim_date = df_dim_date.dropDuplicates(["date_id"])
print(f"dim_date: {df_dim_date.count()}")

df_customers = df_raw.select(
    col("sale_customer_id").cast("int").alias("customer_id"),
    col("customer_first_name").cast("string").alias("first_name"),
    col("customer_last_name").cast("string").alias("last_name"),
    col("customer_age").cast("int").alias("age"),
    col("customer_email").cast("string").alias("email"),
    col("customer_country").cast("string").alias("country"),
    col("customer_postal_code").cast("string").alias("postal_code"),
    col("customer_pet_type").cast("string").alias("pet_type"),
    col("customer_pet_name").cast("string").alias("pet_name"),
    col("customer_pet_breed").cast("string").alias("pet_breed")
).distinct().na.drop(subset=["customer_id"])

df_customers = df_customers.dropDuplicates(["customer_id"])
print(f"dim_customer: {df_customers.count()}")

df_sellers = df_raw.select(
    col("sale_seller_id").cast("int").alias("seller_id"),
    col("seller_first_name").cast("string").alias("first_name"),
    col("seller_last_name").cast("string").alias("last_name"),
    col("seller_email").cast("string").alias("email"),
    col("seller_country").cast("string").alias("country"),
    col("seller_postal_code").cast("string").alias("postal_code")
).distinct().na.drop(subset=["seller_id"])

df_sellers = df_sellers.dropDuplicates(["seller_id"])
print(f"dim_seller: {df_sellers.count()}")

df_suppliers_raw = df_raw.select(
    col("supplier_name").cast("string"),
    col("supplier_contact").cast("string"),
    col("supplier_email").cast("string"),
    col("supplier_phone").cast("string"),
    col("supplier_address").cast("string"),
    col("supplier_city").cast("string"),
    col("supplier_country").cast("string")
).distinct().na.drop(subset=["supplier_name"])

df_suppliers_raw = df_suppliers_raw.dropDuplicates(["supplier_name"])

window_supplier = Window.orderBy("supplier_name")
df_dim_supplier = df_suppliers_raw.withColumn("supplier_id", row_number().over(window_supplier))
print(f"dim_supplier: {df_dim_supplier.count()}")

df_products_raw = df_raw.select(
    col("sale_product_id").cast("int").alias("product_id"),
    col("product_name").cast("string"),
    col("product_category").cast("string"),
    col("pet_category").cast("string"),
    col("product_price").cast("float"),
    col("product_weight").cast("float"),
    col("product_color").cast("string"),
    col("product_size").cast("string"),
    col("product_brand").cast("string"),
    col("product_material").cast("string"),
    col("product_description").cast("string"),
    col("product_rating").cast("float"),
    col("product_reviews").cast("int"),
    col("product_release_date"),
    col("product_expiry_date"),
    col("supplier_name").cast("string")
).distinct().na.drop(subset=["product_id"])

df_dim_product = df_products_raw.join(df_dim_supplier, on="supplier_name", how="left") \
    .drop("supplier_name") \
    .select("product_id", "product_name", "product_category", "pet_category", "product_price",
            "product_weight", "product_color", "product_size", "product_brand", "product_material",
            "product_description", "product_rating", "product_reviews",
            "product_release_date", "product_expiry_date", "supplier_id")
df_dim_product = df_dim_product.dropDuplicates(["product_id"])
print(f"dim_product: {df_dim_product.count()}")

df_stores_raw = df_raw.select(
    col("store_name").cast("string"),
    col("store_location").cast("string"),
    col("store_city").cast("string"),
    col("store_state").cast("string"),
    col("store_country").cast("string"),
    col("store_phone").cast("string"),
    col("store_email").cast("string")
).distinct().na.drop(subset=["store_name"])

df_stores_raw = df_stores_raw.dropDuplicates(["store_name"])

window_store = Window.orderBy("store_name")
df_dim_store = df_stores_raw.withColumn("store_id", row_number().over(window_store))
df_dim_store = df_dim_store.dropDuplicates(["store_id", "store_name"])
print(f"dim_store: {df_dim_store.count()}")

df_fact_base = df_raw.filter(col("sale_date").isNotNull()) \
    .select("sale_date", "sale_customer_id", "sale_seller_id", "sale_product_id",
            "store_name", "sale_quantity", "sale_total_price", "product_price")

dim_date_keys = df_dim_date.select("full_date", "date_id").distinct()
dim_customer_keys = df_customers.select("customer_id").distinct()
dim_seller_keys = df_sellers.select("seller_id").distinct()
dim_product_keys = df_dim_product.select("product_id").distinct()
dim_store_keys = df_dim_store.select("store_id", "store_name").distinct()
```

```python
df_fact = df_fact_base.join(dim_date_keys, 
                            df_fact_base["sale_date"] == dim_date_keys["full_date"], 
                            "left") \
                      .drop("sale_date", "full_date")

df_fact = df_fact.join(dim_customer_keys, 
                       df_fact["sale_customer_id"] == dim_customer_keys["customer_id"], 
                       "left") \
                 .drop("sale_customer_id")

df_fact = df_fact.join(dim_seller_keys, 
                       df_fact["sale_seller_id"] == dim_seller_keys["seller_id"], 
                       "left") \
                 .drop("sale_seller_id")

df_fact = df_fact.join(dim_product_keys, 
                       df_fact["sale_product_id"] == dim_product_keys["product_id"], 
                       "left") \
                 .drop("sale_product_id")

df_fact = df_fact.join(dim_store_keys, 
                       df_fact["store_name"] == dim_store_keys["store_name"], 
                       "left") \
                 .drop("store_name")

df_fact = df_fact.withColumnRenamed("sale_quantity", "quantity") \
                 .withColumnRenamed("sale_total_price", "total_price") \
                 .withColumnRenamed("product_price", "unit_price") \
                 .select("date_id", "customer_id", "seller_id", "product_id", "store_id",
                         "quantity", "total_price", "unit_price")

df_fact = df_fact.na.drop(subset=["date_id", "customer_id", "seller_id", "product_id", "store_id"])
print(f"fact_sales: {df_fact.count()}")
df_fact.show(5, False)
```

```python
def write_to_postgres(df, table_name):
    df.write.jdbc(url=JDBC_URL, table=f"star.{table_name}", mode="append", properties=DB_PROPS)

conn = psycopg2.connect(host="postgres", port=5432, dbname="postgres", 
                        user="postgres", password="POSTGRES_PASSWORD")
conn.autocommit = True
cur = conn.cursor()
cur.execute("SET session_replication_role = 'replica';")
for t in ["fact_sales", "dim_product", "dim_customer", "dim_seller", "dim_supplier", "dim_store", "dim_date"]:
    cur.execute(f"TRUNCATE TABLE star.{t} CASCADE;")
cur.execute("SET session_replication_role = 'origin';")
cur.close()
conn.close()
print("Таблицы очищены.")

print("Запись dim_date...")
write_to_postgres(df_dim_date, "dim_date")
print("Запись dim_customer...")
write_to_postgres(df_customers, "dim_customer")
print("Запись dim_seller...")
write_to_postgres(df_sellers, "dim_seller")
print("Запись dim_supplier...")
write_to_postgres(df_dim_supplier, "dim_supplier")
print("Запись dim_product...")
write_to_postgres(df_dim_product, "dim_product")
print("Запись dim_store...")
write_to_postgres(df_dim_store, "dim_store")
print("Запись fact_sales...")
write_to_postgres(df_fact, "fact_sales")
print("ETL успешно завершён!")

df_check = spark.read.jdbc(url=JDBC_URL, table="star.fact_sales", properties=DB_PROPS)
print("Количество записей в fact_sales:", df_check.count())
df_check.groupBy("date_id").count().orderBy("date_id").show(5)

spark.stop()
```
